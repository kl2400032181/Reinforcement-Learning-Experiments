```python
# Create Grid World

grid = [
    ['S', '.', '.', '.'],
    ['.', '.', '.', '.'],
    ['.', '.', '.', '.'],
    ['.', '.', '.', 'G']
]

for row in grid:
    print(row)
```

```python
# Simulate agent movement

position = [0, 0]

actions = ['Right', 'Right', 'Down', 'Down', 'Down', 'Right']

for action in actions:

    if action == 'Right':
        position[1] += 1

    elif action == 'Down':
        position[0] += 1

    print("Action:", action, "Position:", position)

print("Goal Reached!")
```

```python
# Observe rewards

rewards = [-1, -1, -1, -1, -1, 10]

total_reward = 0

for reward in rewards:
    print("Reward:", reward)
    total_reward += reward

print("Total Reward:", total_reward)
```
