```python
import numpy as np

n = 4
gamma = 0.9

actions = [(-1, 0), (1, 0), (0, -1), (0, 1)]
action_names = ["↑", "↓", "←", "→"]

terminal = (3, 3)

# Initial state values
V = np.zeros((n, n))


def get_next_state(i, j, action):

    ni = i + action[0]
    nj = j + action[1]

    ni = max(0, min(n - 1, ni))
    nj = max(0, min(n - 1, nj))

    return ni, nj


# Value Iteration
for iteration in range(100):

    new_V = V.copy()

    for i in range(n):
        for j in range(n):

            if (i, j) == terminal:
                continue

            action_values = []

            for action in actions:

                ni, nj = get_next_state(i, j, action)

                reward = 1 if (ni, nj) == terminal else 0

                value = reward + gamma * V[ni][nj]

                action_values.append(value)

            new_V[i][j] = max(action_values)

    # Check convergence
    if np.max(np.abs(new_V - V)) < 0.001:
        V = new_V
        break

    V = new_V


# Extract optimal policy
policy = np.zeros((n, n), dtype=int)

for i in range(n):
    for j in range(n):

        if (i, j) == terminal:
            continue

        action_values = []

        for action in actions:

            ni, nj = get_next_state(i, j, action)

            reward = 1 if (ni, nj) == terminal else 0

            value = reward + gamma * V[ni][nj]

            action_values.append(value)

        policy[i][j] = np.argmax(action_values)


print("Optimal State Values:")
print(V)

print("\nOptimal Policy:")

for i in range(n):
    for j in range(n):

        if (i, j) == terminal:
            print("G", end=" ")
        else:
            print(action_names[policy[i][j]], end=" ")

    print()

print("\nNumber of Iterations:", iteration + 1)
```
