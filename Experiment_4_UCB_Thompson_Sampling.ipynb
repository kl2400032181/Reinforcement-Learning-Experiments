```python
import math
import random

k = 3
n = 10

q = [0, 0, 0]       # Average rewards
count = [0, 0, 0]   # Number of times actions selected

total_reward = 0

for t in range(1, n + 1):

    # Select each action once
    if t <= k:
        action = t - 1

    else:
        ucb = []

        for i in range(k):
            value = q[i] + math.sqrt(
                (2 * math.log(t)) / count[i]
            )
            ucb.append(value)

        action = ucb.index(max(ucb))

    # Get random reward
    reward = random.random()

    # Update count
    count[action] += 1

    # Update average reward
    q[action] += (reward - q[action]) / count[action]

    total_reward += reward

print("Average Rewards:", q)
print("Action Counts:", count)
print("Total Reward:", total_reward)
```

```python
import random

k = 3
n = 10

alpha = [1, 1, 1]
beta = [1, 1, 1]

total_reward = 0

for i in range(n):

    samples = []

    # Take a sample for each action
    for j in range(k):
        sample = random.betavariate(alpha[j], beta[j])
        samples.append(sample)

    # Select highest sample
    action = samples.index(max(samples))

    # Get reward
    reward = random.randint(0, 1)

    # Update
    if reward == 1:
        alpha[action] += 1
    else:
        beta[action] += 1

    total_reward += reward

print("Alpha:", alpha)
print("Beta:", beta)
print("Total Reward:", total_reward)
```
